```python
# ============================================================
# MISSING VALUE ASSESSMENT
# ============================================================

missing_value_summary = []

for name, df in datasets.items():

    for column in df.columns:

        missing_count = df[column].isna().sum()

        missing_value_summary.append({
            "Dataset": name,
            "Column": column,
            "Missing Values": missing_count,
            "Missing Percentage": round(
                (missing_count / len(df)) * 100, 2
            ) if len(df) > 0 else 0
        })

missing_summary = pd.DataFrame(missing_value_summary)

missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
].sort_values(
    by=["Dataset", "Missing Percentage"],
    ascending=[True, False]
).reset_index(drop=True)

if missing_summary.empty:
    print("No missing values detected.")
else:
    display(missing_summary)
```

```python
# ============================================================
# DUPLICATE RECORD ASSESSMENT
# ============================================================

duplicate_summary = []

for name, df in datasets.items():

    duplicate_summary.append({
        "Dataset": name,
        "Total Records": len(df),
        "Duplicate Rows": df.duplicated().sum(),
        "Duplicate Percentage": round(
            (df.duplicated().sum() / len(df)) * 100, 2
        ) if len(df) > 0 else 0
    })

duplicate_summary = pd.DataFrame(duplicate_summary)

display(duplicate_summary)
```

```python
# ============================================================
# UNIQUE VALUE ASSESSMENT
# ============================================================

for dataset_name, columns in {
    "encounters": [
        "ENCOUNTERCLASS",
        "PAYER",
        "ORGANIZATION"
    ],
    "patients": [
        "GENDER",
        "MARITAL",
        "RACE",
        "ETHNICITY"
    ],
    "procedures": [
        "CODE",
        "DESCRIPTION"
    ],
    "payers": [
        "NAME"
    ],
    "organizations": [
        "NAME"
    ]
}.items():

    df = datasets[dataset_name]

    print("\n" + "=" * 60)
    print(f"DATASET: {dataset_name.upper()}")
    print("=" * 60)

    for column in columns:

        print(f"\n{column}")
        print("Unique values:", df[column].nunique(dropna=True))
```

```python
# ============================================================
# PRIMARY KEY VALIDATION
# ============================================================

primary_key_checks = {
    "encounters": "Id",
    "patients": "Id",
    "payers": "Id",
    "organizations": "Id"
}

primary_key_results = []

for dataset_name, key_column in primary_key_checks.items():

    df = datasets[dataset_name]

    primary_key_results.append({
        "Dataset": dataset_name,
        "Key": key_column,
        "Total Records": len(df),
        "Missing Keys": df[key_column].isna().sum(),
        "Unique Keys": df[key_column].nunique(),
        "Duplicate Keys": df[key_column].duplicated().sum()
    })

primary_key_summary = pd.DataFrame(primary_key_results)

display(primary_key_summary)
```

```python
# ============================================================
# FOREIGN KEY VALIDATION
# ============================================================

relationship_checks = [
    ("encounters", "PATIENT", "patients", "Id"),
    ("encounters", "PAYER", "payers", "Id"),
    ("encounters", "ORGANIZATION", "organizations", "Id"),
    ("procedures", "PATIENT", "patients", "Id"),
    ("procedures", "ENCOUNTER", "encounters", "Id")
]

relationship_results = []

for child_name, child_column, parent_name, parent_column in relationship_checks:

    child_df = datasets[child_name]
    parent_df = datasets[parent_name]

    child_values = child_df[child_column].dropna()
    parent_values = set(parent_df[parent_column].dropna())

    unmatched = child_values[~child_values.isin(parent_values)]

    relationship_results.append({
        "Relationship": f"{child_name}.{child_column} → {parent_name}.{parent_column}",
        "Records Checked": len(child_values),
        "Unmatched References": len(unmatched),
        "Unmatched Percentage": round(
            len(unmatched) / len(child_values) * 100, 2
        ) if len(child_values) else 0
    })

relationship_summary = pd.DataFrame(relationship_results)

display(relationship_summary)
```

```python
# ============================================================
# DATE FIELD VALIDATION
# ============================================================

date_columns = {
    "encounters": ["START", "STOP"],
    "patients": ["BIRTHDATE", "DEATHDATE"],
    "procedures": ["START", "STOP"]
}

date_results = []

for dataset_name, columns in date_columns.items():

    df = datasets[dataset_name]

    for column in columns:

        original_values = df[column]
        parsed_dates = pd.to_datetime(
            original_values,
            errors="coerce",
            utc=True
        )

        invalid_dates = (
            original_values.notna() & parsed_dates.isna()
        ).sum()

        date_results.append({
            "Dataset": dataset_name,
            "Column": column,
            "Non-Missing Values": original_values.notna().sum(),
            "Unparseable Dates": invalid_dates,
            "Earliest Date": parsed_dates.min(),
            "Latest Date": parsed_dates.max()
        })

date_summary = pd.DataFrame(date_results)

display(date_summary)
```

```python
# ============================================================
# PAYER MISSING-VALUE INVESTIGATION
# ============================================================

payers = datasets["payers"]

payer_missing = payers[payers.isna().any(axis=1)]

display(
    payer_missing[["Id", "NAME"]]
)

print("Payer records with at least one missing field:",
      len(payer_missing))
```

```python
# ============================================================
# CHRONOLOGICAL VALIDITY CHECK
# ============================================================

date_pair_checks = {
    "encounters": ("START", "STOP"),
    "procedures": ("START", "STOP")
}

chronology_results = []

for dataset_name, (start_col, stop_col) in date_pair_checks.items():

    df = datasets[dataset_name]

    start = pd.to_datetime(df[start_col], errors="coerce", utc=True)
    stop = pd.to_datetime(df[stop_col], errors="coerce", utc=True)

    chronology_results.append({
        "Dataset": dataset_name,
        "Total Records": len(df),
        "Start After Stop": (start > stop).sum(),
        "Start Equals Stop": (start == stop).sum(),
        "Valid Positive Duration": (start < stop).sum(),
        "Missing Start or Stop": (start.isna() | stop.isna()).sum()
    })

chronology_summary = pd.DataFrame(chronology_results)

display(chronology_summary)
```

```python
# ============================================================
# PROCEDURE-ENCOUNTER PATIENT CONSISTENCY
# ============================================================

procedures = datasets["procedures"]
encounters = datasets["encounters"]

patient_comparison = procedures.merge(
    encounters[["Id", "PATIENT"]],
    left_on="ENCOUNTER",
    right_on="Id",
    how="left",
    suffixes=("_PROCEDURE", "_ENCOUNTER")
)

patient_mismatches = (
    patient_comparison["PATIENT_PROCEDURE"] !=
    patient_comparison["PATIENT_ENCOUNTER"]
)

print("Total procedures:", len(patient_comparison))
print("Patient mismatches:", patient_mismatches.sum())
print(
    "Mismatch percentage:",
    round(patient_mismatches.mean() * 100, 2), "%"
)
```

```python
# ============================================================
# ENCOUNTER DURATION ANALYSIS
# ============================================================

encounters = datasets["encounters"].copy()

encounters["START_DATETIME"] = pd.to_datetime(
    encounters["START"], errors="coerce", utc=True
)

encounters["STOP_DATETIME"] = pd.to_datetime(
    encounters["STOP"], errors="coerce", utc=True
)

encounters["DURATION_HOURS"] = (
    encounters["STOP_DATETIME"] -
    encounters["START_DATETIME"]
).dt.total_seconds() / 3600

valid_durations = encounters["DURATION_HOURS"].dropna()

long_encounters = valid_durations > 24

duration_summary = pd.DataFrame({
    "Metric": [
        "Total encounters",
        "Valid durations",
        "Encounters exceeding 24 hours",
        "Encounters of 24 hours or less",
        "Average duration (hours)",
        "Median duration (hours)",
        "Maximum duration (hours)"
    ],
    "Value": [
        len(encounters),
        len(valid_durations),
        long_encounters.sum(),
        (valid_durations <= 24).sum(),
        round(valid_durations.mean(), 2),
        round(valid_durations.median(), 2),
        round(valid_durations.max(), 2)
    ]
})

display(duration_summary)

print(
    "Percentage exceeding 24 hours:",
    round(long_encounters.mean() * 100, 2), "%"
)
```

```python
# Create a working copy of the encounters dataset
encounters = datasets["encounters"].copy()

# Convert timestamps
encounters["START_DATETIME"] = pd.to_datetime(
    encounters["START"], errors="coerce", utc=True
)

encounters["STOP_DATETIME"] = pd.to_datetime(
    encounters["STOP"], errors="coerce", utc=True
)

# Calculate duration in hours
encounters["DURATION_HOURS"] = (
    encounters["STOP_DATETIME"] - encounters["START_DATETIME"]
).dt.total_seconds() / 3600

# Examine the 10 longest encounters
longest_encounters = (
    encounters[
        [
            "Id",
            "START_DATETIME",
            "STOP_DATETIME",
            "ENCOUNTERCLASS",
            "DURATION_HOURS"
        ]
    ]
    .sort_values("DURATION_HOURS", ascending=False)
    .head(10)
)

display(longest_encounters)
```

```python
# Summarise encounters exceeding 24 hours by encounter class
long_encounter_summary = (
    encounters[encounters["DURATION_HOURS"] > 24]
    .groupby("ENCOUNTERCLASS")
    .agg(
        Encounter_Count=("Id", "count"),
        Average_Duration_Hours=("DURATION_HOURS", "mean"),
        Maximum_Duration_Hours=("DURATION_HOURS", "max")
    )
    .reset_index()
)

long_encounter_summary[
    ["Average_Duration_Hours", "Maximum_Duration_Hours"]
] = long_encounter_summary[
    ["Average_Duration_Hours", "Maximum_Duration_Hours"]
].round(2)

display(long_encounter_summary)
```

```python
# Summarise duration distribution by encounter class

duration_by_class = (
    encounters.groupby("ENCOUNTERCLASS")
    .agg(
        Encounter_Count=("Id", "count"),
        Average_Duration_Hours=("DURATION_HOURS", "mean"),
        Median_Duration_Hours=("DURATION_HOURS", "median"),
        Minimum_Duration_Hours=("DURATION_HOURS", "min"),
        Maximum_Duration_Hours=("DURATION_HOURS", "max"),
        Encounters_Over_24_Hours=(
            "DURATION_HOURS",
            lambda x: (x > 24).sum()
        )
    )
    .reset_index()
)

duration_columns = [
    "Average_Duration_Hours",
    "Median_Duration_Hours",
    "Minimum_Duration_Hours",
    "Maximum_Duration_Hours"
]

duration_by_class[duration_columns] = (
    duration_by_class[duration_columns].round(2)
)

duration_by_class["Percentage_Over_24_Hours"] = (
    duration_by_class["Encounters_Over_24_Hours"]
    / duration_by_class["Encounter_Count"] * 100
).round(2)

display(duration_by_class)
```

```python
# Aggregate procedure costs by encounter

procedures = datasets["procedures"].copy()
encounters = datasets["encounters"].copy()

procedure_costs = (
    procedures.groupby("ENCOUNTER")
    .agg(
        Procedure_Count=("CODE", "count"),
        Total_Procedure_Base_Cost=("BASE_COST", "sum")
    )
    .reset_index()
)

# Join procedure cost summaries to encounters
encounter_cost_analysis = encounters.merge(
    procedure_costs,
    left_on="Id",
    right_on="ENCOUNTER",
    how="left"
)

# Encounters without recorded procedures
encounter_cost_analysis["Procedure_Count"] = (
    encounter_cost_analysis["Procedure_Count"].fillna(0)
)

encounter_cost_analysis["Total_Procedure_Base_Cost"] = (
    encounter_cost_analysis["Total_Procedure_Base_Cost"].fillna(0)
)

# Compare recorded claim cost with encounter and procedure base costs
encounter_cost_analysis["Combined_Base_Cost"] = (
    encounter_cost_analysis["BASE_ENCOUNTER_COST"]
    + encounter_cost_analysis["Total_Procedure_Base_Cost"]
)

encounter_cost_analysis["Cost_Difference"] = (
    encounter_cost_analysis["TOTAL_CLAIM_COST"]
    - encounter_cost_analysis["Combined_Base_Cost"]
)

display(
    encounter_cost_analysis[
        [
            "ENCOUNTERCLASS",
            "BASE_ENCOUNTER_COST",
            "Procedure_Count",
            "Total_Procedure_Base_Cost",
            "TOTAL_CLAIM_COST",
            "Combined_Base_Cost",
            "Cost_Difference"
        ]
    ]
    .sort_values("TOTAL_CLAIM_COST", ascending=False)
    .head(10)
    .round(2)
)
```

```python
# Overall comparison of claim costs and recorded base costs

cost_reconciliation = pd.DataFrame({
    "Metric": [
        "Total encounters",
        "Encounters with procedures",
        "Encounters without procedures",
        "Average procedure count per encounter",
        "Average total procedure base cost",
        "Average combined base cost",
        "Average total claim cost",
        "Average difference"
    ],
    "Value": [
        len(encounter_cost_analysis),
        (encounter_cost_analysis["Procedure_Count"] > 0).sum(),
        (encounter_cost_analysis["Procedure_Count"] == 0).sum(),
        encounter_cost_analysis["Procedure_Count"].mean(),
        encounter_cost_analysis["Total_Procedure_Base_Cost"].mean(),
        encounter_cost_analysis["Combined_Base_Cost"].mean(),
        encounter_cost_analysis["TOTAL_CLAIM_COST"].mean(),
        encounter_cost_analysis["Cost_Difference"].mean()
    ]
})

cost_reconciliation["Value"] = cost_reconciliation["Value"].round(2)

display(cost_reconciliation)
```

```python
# Check whether each procedure code maps to one description

code_description_counts = (
    procedures.groupby("CODE")["DESCRIPTION"]
    .nunique()
    .reset_index(name="Description_Count")
)

codes_with_multiple_descriptions = code_description_counts[
    code_description_counts["Description_Count"] > 1
]

# Check whether descriptions map to multiple codes

description_code_counts = (
    procedures.groupby("DESCRIPTION")["CODE"]
    .nunique()
    .reset_index(name="Code_Count")
)

descriptions_with_multiple_codes = description_code_counts[
    description_code_counts["Code_Count"] > 1
]

print(
    "Procedure codes with multiple descriptions:",
    len(codes_with_multiple_descriptions)
)

print(
    "Procedure descriptions with multiple codes:",
    len(descriptions_with_multiple_codes)
)

display(codes_with_multiple_descriptions.head(10))
display(descriptions_with_multiple_codes.head(10))
```

```python
# Join payer names to encounter records

payers = datasets["payers"].copy()
encounters = datasets["encounters"].copy()

payer_analysis = encounters.merge(
    payers[["Id", "NAME"]],
    left_on="PAYER",
    right_on="Id",
    how="left"
)

# Summarise encounters by payer
payer_summary = (
    payer_analysis.groupby("NAME")
    .agg(
        Encounter_Count=("Id_x", "count"),
        Average_Claim_Cost=("TOTAL_CLAIM_COST", "mean"),
        Median_Claim_Cost=("TOTAL_CLAIM_COST", "median"),
        Total_Claim_Cost=("TOTAL_CLAIM_COST", "sum"),
        Average_Payer_Coverage=("PAYER_COVERAGE", "mean"),
        Total_Payer_Coverage=("PAYER_COVERAGE", "sum"),
        Zero_Coverage_Encounters=(
            "PAYER_COVERAGE",
            lambda x: (x == 0).sum()
        )
    )
    .reset_index()
    .sort_values("Total_Claim_Cost", ascending=False)
)

display(payer_summary.round(2))
```

```python
# Calculate payer coverage as a proportion of total claim costs

payer_summary["Coverage_Percentage"] = (
    payer_summary["Total_Payer_Coverage"]
    / payer_summary["Total_Claim_Cost"] * 100
).round(2)

payer_summary["Zero_Coverage_Percentage"] = (
    payer_summary["Zero_Coverage_Encounters"]
    / payer_summary["Encounter_Count"] * 100
).round(2)

display(
    payer_summary[
        [
            "NAME",
            "Encounter_Count",
            "Total_Claim_Cost",
            "Total_Payer_Coverage",
            "Coverage_Percentage",
            "Zero_Coverage_Encounters",
            "Zero_Coverage_Percentage"
        ]
    ].round(2)
)
```

```python
# Investigate positive and zero payer coverage by payer

payer_coverage_check = (
    payer_analysis
    .assign(
        Has_Positive_Coverage=lambda df: df["PAYER_COVERAGE"] > 0
    )
    .groupby("NAME")
    .agg(
        Encounter_Count=("Id_x", "count"),
        Zero_Coverage_Count=(
            "PAYER_COVERAGE",
            lambda x: (x == 0).sum()
        ),
        Positive_Coverage_Count=(
            "Has_Positive_Coverage",
            "sum"
        ),
        Total_Payer_Coverage=("PAYER_COVERAGE", "sum"),
        Average_Positive_Coverage=(
            "PAYER_COVERAGE",
            lambda x: x[x > 0].mean()
        ),
        Median_Positive_Coverage=(
            "PAYER_COVERAGE",
            lambda x: x[x > 0].median()
        ),
        Maximum_Payer_Coverage=("PAYER_COVERAGE", "max")
    )
    .reset_index()
    .sort_values("Total_Payer_Coverage", ascending=False)
)

display(payer_coverage_check.round(2))
```

```python
# Validate payer-level aggregation against the original encounters

print("Original encounter count:", len(encounters))
print("Payer summary encounter count:", payer_summary["Encounter_Count"].sum())

print("\nOriginal total claim cost:",
      round(encounters["TOTAL_CLAIM_COST"].sum(), 2))

print("Payer summary total claim cost:",
      round(payer_summary["Total_Claim_Cost"].sum(), 2))

print("\nOriginal total payer coverage:",
      round(encounters["PAYER_COVERAGE"].sum(), 2))

print("Payer summary total payer coverage:",
      round(payer_summary["Total_Payer_Coverage"].sum(), 2))
```

## Stage 2 — Analytical Data Modelling

This stage prepares the encounter and procedure fact tables
and the associated dimension tables.

The raw datasets will remain unchanged. All transformations
will be performed on copies of the source data.

The model will preserve the different grains of encounter-level
and procedure-level data to prevent duplication of encounter
financial values during analysis.

```python
# Create independent copies of the source datasets

encounters = datasets["encounters"].copy()
procedures = datasets["procedures"].copy()
patients = datasets["patients"].copy()
payers = datasets["payers"].copy()
organizations = datasets["organizations"].copy()

print("Analytical source copies created.")

print("Encounters:", encounters.shape)
print("Procedures:", procedures.shape)
print("Patients:", patients.shape)
print("Payers:", payers.shape)
print("Organizations:", organizations.shape)
```

```python
# Prepare the encounter fact table

fact_encounters = encounters[
    [
        "Id",
        "PATIENT",
        "PAYER",
        "ORGANIZATION",
        "ENCOUNTERCLASS",
        "START",
        "STOP",
        "BASE_ENCOUNTER_COST",
        "TOTAL_CLAIM_COST",
        "PAYER_COVERAGE"
    ]
].copy()

# Convert timestamps
fact_encounters["START"] = pd.to_datetime(
    fact_encounters["START"], utc=True
)

fact_encounters["STOP"] = pd.to_datetime(
    fact_encounters["STOP"], utc=True
)

# Derive duration in hours
fact_encounters["DURATION_HOURS"] = (
    fact_encounters["STOP"] - fact_encounters["START"]
).dt.total_seconds() / 3600

# Derive date fields
fact_encounters["START_DATE"] = (
    fact_encounters["START"].dt.date
)

fact_encounters["STOP_DATE"] = (
    fact_encounters["STOP"].dt.date
)

# Flag encounters exceeding 24 hours
fact_encounters["OVER_24_HOURS"] = (
    fact_encounters["DURATION_HOURS"] > 24
)

display(fact_encounters.head())
```

```python
# Validate the encounter fact table

print("Original encounter records:", len(encounters))
print("Fact encounter records:", len(fact_encounters))

print(
    "Unique encounter IDs:",
    fact_encounters["Id"].nunique()
)

print(
    "Missing encounter IDs:",
    fact_encounters["Id"].isna().sum()
)

print(
    "Duplicate encounter IDs:",
    fact_encounters["Id"].duplicated().sum()
)

print(
    "Missing duration values:",
    fact_encounters["DURATION_HOURS"].isna().sum()
)

print(
    "Encounters exceeding 24 hours:",
    fact_encounters["OVER_24_HOURS"].sum()
)
```

```python
# Prepare the procedure fact table

fact_procedures = procedures[
    [
        "PATIENT",
        "ENCOUNTER",
        "START",
        "STOP",
        "CODE",
        "DESCRIPTION",
        "BASE_COST"
    ]
].copy()

# Create a unique identifier for each procedure record
fact_procedures.insert(
    0,
    "PROCEDURE_ID",
    range(1, len(fact_procedures) + 1)
)

# Convert timestamps
fact_procedures["START"] = pd.to_datetime(
    fact_procedures["START"], utc=True
)

fact_procedures["STOP"] = pd.to_datetime(
    fact_procedures["STOP"], utc=True
)

# Calculate procedure duration in hours
fact_procedures["DURATION_HOURS"] = (
    fact_procedures["STOP"] - fact_procedures["START"]
).dt.total_seconds() / 3600

# Derive procedure date
fact_procedures["PROCEDURE_DATE"] = (
    fact_procedures["START"].dt.date
)

# Flag zero-duration procedures
fact_procedures["ZERO_DURATION"] = (
    fact_procedures["DURATION_HOURS"] == 0
)

display(fact_procedures.head())
```

```python
# Validate the procedure fact table

print("Original procedure records:", len(procedures))
print("Fact procedure records:", len(fact_procedures))

print(
    "Unique procedure IDs:",
    fact_procedures["PROCEDURE_ID"].nunique()
)

print(
    "Missing procedure IDs:",
    fact_procedures["PROCEDURE_ID"].isna().sum()
)

print(
    "Duplicate procedure IDs:",
    fact_procedures["PROCEDURE_ID"].duplicated().sum()
)

print(
    "Missing encounter references:",
    fact_procedures["ENCOUNTER"].isna().sum()
)

print(
    "Unmatched encounter references:",
    (~fact_procedures["ENCOUNTER"].isin(
        fact_encounters["Id"]
    )).sum()
)

print(
    "Missing duration values:",
    fact_procedures["DURATION_HOURS"].isna().sum()
)

print(
    "Zero-duration procedures:",
    fact_procedures["ZERO_DURATION"].sum()
)
```

```python
# Select encounter attributes required for procedure analysis

encounter_attributes = fact_encounters[
    [
        "Id",
        "PATIENT",
        "PAYER",
        "ORGANIZATION",
        "ENCOUNTERCLASS",
        "START_DATE"
    ]
].copy()

# Rename fields to distinguish encounter information
encounter_attributes = encounter_attributes.rename(
    columns={
        "Id": "ENCOUNTER",
        "PATIENT": "ENCOUNTER_PATIENT",
        "START_DATE": "ENCOUNTER_DATE"
    }
)

# Enrich the procedure fact table
fact_procedures = fact_procedures.merge(
    encounter_attributes,
    on="ENCOUNTER",
    how="left",
    validate="many_to_one"
)

display(fact_procedures.head())
```

```python
# Validate the enriched procedure fact table

print("Procedure records:", len(fact_procedures))

print(
    "Unique procedure IDs:",
    fact_procedures["PROCEDURE_ID"].nunique()
)

print(
    "Missing payer references:",
    fact_procedures["PAYER"].isna().sum()
)

print(
    "Missing organisation references:",
    fact_procedures["ORGANIZATION"].isna().sum()
)

print(
    "Missing encounter classes:",
    fact_procedures["ENCOUNTERCLASS"].isna().sum()
)

print(
    "Missing encounter dates:",
    fact_procedures["ENCOUNTER_DATE"].isna().sum()
)

print(
    "Patient reference mismatches:",
    (
        fact_procedures["PATIENT"]
        != fact_procedures["ENCOUNTER_PATIENT"]
    ).sum()
)

print(
    "Duplicate procedure IDs:",
    fact_procedures["PROCEDURE_ID"].duplicated().sum()
)
```

```python
# 1. Patient dimension
dim_patients = (
    patients[["Id"]]
    .drop_duplicates()
    .rename(columns={"Id": "PATIENT"})
    .reset_index(drop=True)
)

# 2. Payer dimension
dim_payers = (
    payers[["Id", "NAME"]]
    .drop_duplicates()
    .rename(columns={
        "Id": "PAYER",
        "NAME": "PAYER_NAME"
    })
    .reset_index(drop=True)
)

# 3. Organisation dimension
dim_organizations = (
    organizations[["Id", "NAME", "CITY", "STATE"]]
    .drop_duplicates()
    .rename(columns={
        "Id": "ORGANIZATION",
        "NAME": "ORGANIZATION_NAME"
    })
    .reset_index(drop=True)
)

# 4. Encounter class dimension
dim_encounter_class = (
    fact_encounters[["ENCOUNTERCLASS"]]
    .drop_duplicates()
    .sort_values("ENCOUNTERCLASS")
    .reset_index(drop=True)
)

# Display dimension sizes
dimension_sizes = pd.DataFrame({
    "Dimension": [
        "DimPatients",
        "DimPayers",
        "DimOrganizations",
        "DimEncounterClass"
    ],
    "Rows": [
        len(dim_patients),
        len(dim_payers),
        len(dim_organizations),
        len(dim_encounter_class)
    ]
})

display(dimension_sizes)
```

```python
# Identify the earliest and latest dates across both fact tables
earliest_date = min(
    fact_encounters["START_DATE"].min(),
    fact_procedures["PROCEDURE_DATE"].min()
)

latest_date = max(
    fact_encounters["START_DATE"].max(),
    fact_procedures["PROCEDURE_DATE"].max()
)

# Generate a continuous calendar
date_range = pd.date_range(
    start=earliest_date,
    end=latest_date,
    freq="D"
)

# Create the date dimension
dim_date = pd.DataFrame({
    "DATE": date_range
})

dim_date["YEAR"] = dim_date["DATE"].dt.year
dim_date["QUARTER"] = "Q" + dim_date["DATE"].dt.quarter.astype(str)
dim_date["MONTH_NUMBER"] = dim_date["DATE"].dt.month
dim_date["MONTH_NAME"] = dim_date["DATE"].dt.month_name()
dim_date["YEAR_MONTH"] = dim_date["DATE"].dt.strftime("%Y-%m")
dim_date["DAY"] = dim_date["DATE"].dt.day
dim_date["DAY_NAME"] = dim_date["DATE"].dt.day_name()

display(dim_date.head())
display(dim_date.tail())

print("Earliest date:", dim_date["DATE"].min())
print("Latest date:", dim_date["DATE"].max())
print("Total calendar dates:", len(dim_date))
```

```python
fact_procedures = fact_procedures.drop(
    columns=["ENCOUNTER_PATIENT"]
)

print("Remaining columns:")
print(fact_procedures.columns.tolist())
```

```python
dimension_checks = {
    "DimPatients": {
        "Rows": len(dim_patients),
        "Unique keys": dim_patients["PATIENT"].nunique(),
        "Missing keys": dim_patients["PATIENT"].isna().sum()
    },
    "DimPayers": {
        "Rows": len(dim_payers),
        "Unique keys": dim_payers["PAYER"].nunique(),
        "Missing keys": dim_payers["PAYER"].isna().sum()
    },
    "DimOrganizations": {
        "Rows": len(dim_organizations),
        "Unique keys": dim_organizations["ORGANIZATION"].nunique(),
        "Missing keys": dim_organizations["ORGANIZATION"].isna().sum()
    },
    "DimEncounterClass": {
        "Rows": len(dim_encounter_class),
        "Unique keys": dim_encounter_class["ENCOUNTERCLASS"].nunique(),
        "Missing keys": dim_encounter_class["ENCOUNTERCLASS"].isna().sum()
    },
    "DimDate": {
        "Rows": len(dim_date),
        "Unique keys": dim_date["DATE"].nunique(),
        "Missing keys": dim_date["DATE"].isna().sum()
    }
}

dimension_validation = pd.DataFrame(dimension_checks).T

display(dimension_validation)
```

```python
# Recreate the procedure fact table from the original source

fact_procedures = procedures[
    [
        "PATIENT",
        "ENCOUNTER",
        "START",
        "STOP",
        "CODE",
        "DESCRIPTION",
        "BASE_COST"
    ]
].copy()

fact_procedures.insert(
    0, "PROCEDURE_ID", range(1, len(fact_procedures) + 1)
)

# Convert dates
fact_procedures["START"] = pd.to_datetime(
    fact_procedures["START"], utc=True
)

fact_procedures["STOP"] = pd.to_datetime(
    fact_procedures["STOP"], utc=True
)

# Derive procedure duration and date
fact_procedures["DURATION_HOURS"] = (
    fact_procedures["STOP"] - fact_procedures["START"]
).dt.total_seconds() / 3600

fact_procedures["PROCEDURE_DATE"] = (
    fact_procedures["START"].dt.date
)

fact_procedures["ZERO_DURATION"] = (
    fact_procedures["DURATION_HOURS"] == 0
)

# Add encounter attributes
fact_procedures = fact_procedures.merge(
    encounter_attributes,
    on="ENCOUNTER",
    how="left",
    validate="many_to_one"
)

# Remove the temporary patient comparison column
fact_procedures = fact_procedures.drop(
    columns=["ENCOUNTER_PATIENT"]
)

# Inspect the final structure
print(fact_procedures.columns.tolist())
print("Procedure records:", len(fact_procedures))
print("Unique procedure IDs:", fact_procedures["PROCEDURE_ID"].nunique())
```

```python
# Define the foreign key relationships to validate

foreign_key_checks = [
    # FactEncounters
    ("FactEncounters", fact_encounters, "PATIENT",
     "DimPatients", dim_patients, "PATIENT"),

    ("FactEncounters", fact_encounters, "PAYER",
     "DimPayers", dim_payers, "PAYER"),

    ("FactEncounters", fact_encounters, "ORGANIZATION",
     "DimOrganizations", dim_organizations, "ORGANIZATION"),

    ("FactEncounters", fact_encounters, "ENCOUNTERCLASS",
     "DimEncounterClass", dim_encounter_class, "ENCOUNTERCLASS"),

    # FactProcedures
    ("FactProcedures", fact_procedures, "PATIENT",
     "DimPatients", dim_patients, "PATIENT"),

    ("FactProcedures", fact_procedures, "PAYER",
     "DimPayers", dim_payers, "PAYER"),

    ("FactProcedures", fact_procedures, "ORGANIZATION",
     "DimOrganizations", dim_organizations, "ORGANIZATION"),

    ("FactProcedures", fact_procedures, "ENCOUNTERCLASS",
     "DimEncounterClass", dim_encounter_class, "ENCOUNTERCLASS")
]

# Validate foreign keys
relationship_results = []

for fact_name, fact_df, fact_key, dim_name, dim_df, dim_key in foreign_key_checks:

    unmatched = ~fact_df[fact_key].isin(dim_df[dim_key])

    relationship_results.append({
        "Fact Table": fact_name,
        "Foreign Key": fact_key,
        "Dimension": dim_name,
        "Unmatched Records": unmatched.sum(),
        "Status": "PASS" if unmatched.sum() == 0 else "CHECK"
    })

relationship_validation = pd.DataFrame(relationship_results)

display(relationship_validation)
```

```python
# Check that every procedure references an existing encounter
unmatched_encounters = ~fact_procedures["ENCOUNTER"].isin(
    fact_encounters["Id"]
)

# Check patient consistency
encounter_patient_lookup = fact_encounters.set_index("Id")["PATIENT"]

expected_patients = fact_procedures["ENCOUNTER"].map(
    encounter_patient_lookup
)

patient_mismatches = (
    fact_procedures["PATIENT"] != expected_patients
)

print("Unmatched encounter references:", unmatched_encounters.sum())
print("Patient reference mismatches:", patient_mismatches.sum())
```

```python
# Convert dates to the same format for comparison
calendar_dates = set(dim_date["DATE"].dt.date)

encounter_dates = set(fact_encounters["START_DATE"])
procedure_dates = set(fact_procedures["PROCEDURE_DATE"])

print(
    "Encounter dates outside calendar:",
    len(encounter_dates - calendar_dates)
)

print(
    "Procedure dates outside calendar:",
    len(procedure_dates - calendar_dates)
)
```
